```python
import re
import random
from datetime import datetime

user_name = ""
mood = ""

jokes = [
    "Why do programmers prefer dark mode? Because light attracts bugs!",
    "Why was the computer cold? It forgot to close Windows.",
    "Why did Python go to school? To improve its class.",
    "I would tell you a UDP joke... but you may not get it."
]

greetings = [
    "Hello!",
    "Hi there!",
    "Hey! Nice to meet you.",
    "Welcome back!"
]

def chatbot_response(user_input):

    global user_name
    global mood

    user_input = user_input.lower().strip()

    if re.search(r"\b(hi|hello|hey|hii)\b", user_input):
        return random.choice(greetings)

    elif "my name is" in user_input:
        user_name = user_input.replace("my name is", "").strip().title()
        return f"Nice to meet you, {user_name}!"

    elif "what is my name" in user_input:
        if user_name:
            return f"Your name is {user_name}."
        else:
            return "I don't know your name yet."

    elif re.search(r"\b(sad|upset|depressed)\b", user_input):
        mood = "sad"
        return "I hope things get better soon. Want to hear a joke?"

    elif re.search(r"\b(happy|great|awesome)\b", user_input):
        mood = "happy"
        return "That's wonderful to hear!"

    elif "your name" in user_input:
        return "I am RuleBot, a rule-based AI chatbot."

    elif "time" in user_input:
        current_time = datetime.now().strftime("%I:%M:%S %p")
        return f"Current time is {current_time}"

    elif "date" in user_input:
        current_date = datetime.now().strftime("%d-%m-%Y")
        return f"Today's date is {current_date}"

    elif "joke" in user_input:
        return random.choice(jokes)

    elif user_input.startswith("calculate"):

        try:
            expression = user_input.replace("calculate", "").strip()
            result = eval(expression)
            return f"Answer = {result}"

        except:
            return "Invalid mathematical expression."

    elif "help" in user_input:
        return """
I can help you with:
1. Greetings
2. Telling jokes
3. Date and time
4. Remembering your name
5. Simple calculations

Examples:
- Hi
- My name is Rahul
- Tell me a joke
- Calculate 5 * 10
"""

    elif "thank" in user_input:
        return "You're welcome!"

    elif re.search(r"\b(exit|bye|quit)\b", user_input):
        return "Goodbye! Have a nice day."

    else:
        unknown_responses = [
            "Sorry, I didn't understand that.",
            "Can you rephrase that?",
            "Interesting... tell me more.",
            "I am still learning."
        ]
        return random.choice(unknown_responses)

print("=" * 50)
print("        RULE-BASED CHATBOT")
print("=" * 50)
print("Type 'bye' to exit.\n")

while True:

    user_input = input("You : ")

    response = chatbot_response(user_input)

    print("Bot :", response)

    if response == "Goodbye! Have a nice day.":
        break
```
