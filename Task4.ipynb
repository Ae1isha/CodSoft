```python
!pip install pandas scikit-learn

# SMART RECOMMENDATION SYSTEM
# Movies + Books + Products
# Category Based Recommendation


import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# DATASET

data = {

    'Item': [

        # Movies
        'Avengers',
        'Batman',
        'Spiderman',

        # Books
        'Harry Potter',
        'The Alchemist',
        'Atomic Habits',

        # Products
        'iPhone 15',
        'Samsung Galaxy',
        'Nike Shoes',
        'Adidas Sneakers'
    ],

    'Category': [

        'Movie',
        'Movie',
        'Movie',

        'Book',
        'Book',
        'Book',

        'Product',
        'Product',
        'Product',
        'Product'
    ],

    'Features': [

        # Movies
        'action superhero marvel adventure',
        'action superhero dc dark',
        'action superhero marvel teenage',

        # Books
        'magic fantasy wizard adventure',
        'motivation life philosophy inspiration',
        'selfhelp productivity habits success',

        # Products
        'smartphone apple ios camera premium',
        'smartphone android samsung camera',
        'sports running fitness shoes',
        'sports casual sneakers fashion'
    ],

    'Rating': [
        9.0,
        8.7,
        8.9,
        9.2,
        8.8,
        9.1,
        9.3,
        8.9,
        8.5,
        8.4
    ]
}

df = pd.DataFrame(data)

# TF-IDF

tfidf = TfidfVectorizer(stop_words='english')

matrix = tfidf.fit_transform(df['Features'])

# COSINE SIMILARITY

similarity = cosine_similarity(matrix)

# RECOMMEND FUNCTION


def recommend(item_name):

    item_name = item_name.lower()

    # Check item exists
    if item_name not in df['Item'].str.lower().values:
        print("Item not found!")
        return

    # Find item index
    item_index = df[df['Item'].str.lower() == item_name].index[0]

    # Get category of selected item
    selected_category = df.iloc[item_index]['Category']

    # Similarity scores
    scores = list(enumerate(similarity[item_index]))

    # Sort scores
    sorted_scores = sorted(scores, key=lambda x: x[1], reverse=True)

    print("\n===================================")
    print("Recommendations for:", item_name.title())
    print("Category:", selected_category)
    print("===================================\n")

    count = 0

    for item in sorted_scores[1:]:

        index = item[0]

        # Recommend only same category
        if df.iloc[index]['Category'] == selected_category:

            print("Item :", df.iloc[index]['Item'])
            print("Features :", df.iloc[index]['Features'])
            print("Rating :", df.iloc[index]['Rating'])
            print("-----------------------------------")

            count += 1

        if count == 5:
            break

# USER INPUT
r̥R̥

item = input("Enter Item Name : ")

recommend(item)
      
```
